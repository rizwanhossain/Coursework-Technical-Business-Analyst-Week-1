```python
import pandas as pd
data = pd.read_csv('docs/recovery_activity_tracker.csv')
```

```python
data.head()
data['activity_type'].unique()
```

```python
average_activity_duration = data.groupby(
    'activity_type')['minutes_spent'].mean()
average_activity_duration
```

```python
frequent_outcomes = data.groupby(
    'outcome_code').count()/data.shape[0]*100
frequent_outcomes
```
